### Introduction
A sigmoid function is a function that has an abrupt change as a function of some variable, meant to represent some kind of regime shift or tipping point as that variable crosses some kind of threshold. An example is shown in Fig. 1.

<p style='text-align: center;'>    
<img src="Sigmafloor.png" />

<strong>Figure 1</strong>. The $\sigma_{floor}(x)$ function.
</p>

The horizontal axis in this figure (the *independent variable*) is labeled "$x$". But what is the meaning of "$x$"? Well, it could be a temperature, in which case we're thinking that some threshold is crossed when the temperature gets above some value. Or $x$ could be a carbon amount in the atmosphere, or even time. The sigmoid function we'll be using here is flexible in the sense that it can accommodate all those possibilities. 

$\sigma_{floor}$ also has three *parameters*, which we'll call $x_{trans}$, $x_{interval}$, and 
$\sigma_{floor,\infty}$. The idea of this CGI is to get you familiar with how each of those controls the shape of $\sigma_{floor}(x)$.

```python
import numpy as np
import matplotlib.pyplot as plt
```

### $\sigma_{floor}$  as a Python function

```python
def sigmafloor(x,x_trans=0,x_interval=1,sigma_floor_infinity=0):
    """Generates a sigmoid (smooth step-down) function with a floor"""
    temp = 1 - 1/(1 + np.exp(-(x-x_trans)*3/x_interval))
    return temp*(1-sigma_floor_infinity)+sigma_floor_infinity
```

### Specifying the independent variable
The cell below specifies an array of temperatures that will be used as the independent variable "$x$".

```python
T = np.linspace(-2,4)
```

### How $\sigma_{floor}(x)$  depends on  $x_{trans}$.
Execute the cell below and take a look a the results.

```python
plt.figure()
plt.plot(T,sigmafloor(T,x_trans=0,x_interval=1,sigma_floor_infinity=0),label='x_trans=0')
plt.plot(T,sigmafloor(T,x_trans=1,x_interval=1,sigma_floor_infinity=0),label='x_trans=1')
plt.plot(T,sigmafloor(T,x_trans=2,x_interval=1,sigma_floor_infinity=0),label='x_trans=2')
plt.grid(True)
plt.xlabel('x')
plt.ylabel('sigma_floor(x)')
plt.legend()
```
