# $CO_2$ Fertilization

## Introduction
As context for this exercise, take a look at the Houghton Carbon Balance diagram reproduced here as Fig. 1. You can see that there's an additional atmosphere-to-carbon flux of $2 \ GtC/yr$, shown in red, on top of the pre-industrial value of $120 \ GtC/yr$. That additional flux constitutes a "negative feedback" in the atmosphere-land carbon exchange, because it tends to offset the extra carbon accumulating in the atmosphere due to human activities. Also known as the "$CO_2$ fertilization effect" (CFE), this negative feedback is the topic of this exercise.

We have a good idea of the physical basis behind the CFE: experiments have shown that if you pump $CO_2$ into a greenhouse, plants in the greenhouse photosynthesize faster. That this is happening on a global scale has also been demonstrated, with the help of measurements of gas exchange deployed at hundreds of ecosystems worldwide in combination with photosynthetic optimization theories (see Chen et al, 2022, https://pmc.ncbi.nlm.nih.gov/articles/PMC8915860/ and Keenan and Williams, 2018, https://www.annualreviews.org/content/journals/10.1146/annurev-environ-102017-030204).

Mathematically, we'll describe CFE on a global scale by expressing the atmosphere-to-land carbon flux as 

$$
F_{al} = k_{al0} +  k_{al1} \times C_{atm} \ \ \ \ (1)
$$

where $C_{atm}$ is the amount of carbon in the atmosphere (usually given in $GtC$ and $k_{al0}$ and $k_{al1}$ are positive constants. Like $mx+b$, this formulation ensures that $F_{al}$ will get bigger in proportion to $C_{atm}$.
