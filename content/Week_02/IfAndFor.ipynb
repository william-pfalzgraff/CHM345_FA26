## "If" blocks
Lots of times in climate modeling, we want to check whether some condition is met before going ahead. Python does this with an *if* block, which looks like

    if (some condition is met):
        line1
        line2
        ...

This is not actually Python code -- it's what we call *pseudo code* because it's supposed to be a little more intuitive than actual code$^*$. The point for now is, the entire set ("block") of indented commands line1, line2, ... will be executed only if that condition is met.

$*$That said, some people refer to Python as "executable pseudo code."


## "For" loops
A different situation arises when we want to do something over and over again, like march our climate model forward over time. That kind of thing calls for a *loop*. Among the many ways to set up a loop, the one we'll explore here is called a *for* loop. It looks a bit like the *if* syntax:

    for (each step in some sequence):
        line1
        line2
        ...

only now, those lines will be executed over and over again -- once for each step in the sequence. Our numerical method for simulating Earth's climate will be to set up a loop like this. We're going to specify exactly how far forward in time we want to go, and *for looping* is well suited for that kind of thing.

## Accumulating
As it turns out, inside our loop, we'll also want to collect values in an array, for plotting later. Sometimes we'll want to accumulate multiple arrays!

### A couple of examples of "for" loops

This is a little different. We're going to create a numpy array first, and loop over it letting "range" figure out how long it is.

### Updating a variable inside a loop
Below is another example of a "for" loop. You can think of it as being in three parts:

- The lines *before* the loop, tagged as "Setting up", get things ready -- there's a variable (my_number) that's going to be getting bigger and bigger as the loop progresses.
- The loop itself. It starts with the word "for", which specifies how many times the indented lines below it will be executed. Each time through, my_number gets 5 added to it -- that's what "my_number = my_number + 5" does: it takes the old value of my_number, adds 5, and stores the result back in my_number.
- A reporting line that executes after the loop is completed. The reporting line isn't indented; if it were, it would be executed every iteration of the loop!

```python
# Setting up
my_number = 0

# Looping
for i in range(6):
    my_number = my_number + 5
    print(my_number)

# Reporting
print("After the loop, my_number=", my_number)
```

### Accumulating
Sometimes, instead of just printing a variable to the screen as we go through a loop, we'd like to accumulate information as we go along. Here we'll do this with a *numpy array*.

We've already met numpy arrays -- that's what np.linspace makes. A numpy array can also be *empty*, which is how we set things up before the loop starts. Then, inside the loop, numbers are appended to it. The key commands in the cell below are

    my_number_array = np.empty(0)

before the loop, which creates an empty array, and

    my_number_array = np.append(my_number_array, my_number)

inside the loop, which tacks the current value of my_number onto the end of the array. Execute the cell below to see how this works. 

```python
# Setting up
my_number = 0

# Starting out with an empty numpy array
my_number_array = np.empty(0)

# Looping
for i in range(6):
    my_number = my_number + 5
    my_number_array = np.append(my_number_array, my_number)
    print(my_number_array)

# Reporting
print("After the loop, my_number_array=", my_number_array)
```

### Examining elements of an array
When an array is short, it's fine to print it out in its entirety, as we've done. But if an array gets too long, there's no point in looking at all that. In such situations, it's often useful to examine the *first* and the *last* element of the array.

In the example below, we create a numpy array, called *my_array*, that contains five numbers. A few notes about this:

1. my_array[0] refers to the *first* element of this array.
1. my_array[4] refers to the *last* element of this array.
1. my_array[-1] refers to the last element of an array.
1. my_array[-2] refers to the second-to-last element of an array.

Why would we be interested in #3? It's just kind of handy -- it allows us to print the last element of an array without actually knowing how long it is. Here's an example:

```python
my_array = np.array([25, 32, 7, 12, 88])
print(my_array)
print("First element:", my_array[0])
print("Last element:", my_array[-1])
print("Second-to-last element:", my_array[-2])
```

### Building arrays of numbers within a loop over time
Next we'll examine how to accumulate values *calculated* within a loop -- in this case, a loop over a time array, like the one we set up earlier. This time we'll accumulate two numpy arrays at once.

We're also taking this opportunity to introduce how to annotate two curves on the same graph using the "label/legend" method (which can be very handy). 

```python
# Create a time array (this one will have 30 elements)
time = np.linspace(0,20,30)

# Initialize a couple of empty numpy arrays
myarray1 = np.empty(0)
myarray2 = np.empty(0)

# Loop over the elements of time
for i in range(len(time)):
    
    # Extract the time in this iteration, and square it
    thisvalue1 = time[i]**2
    
    # Make another value that's twice what we just got
    thisvalue2 = 2*thisvalue1
    
    # append both of these values to our arrays
    myarray1 = np.append(myarray1,thisvalue1)
    myarray2 = np.append(myarray2,thisvalue2)

# Graph the results using the label/legend method
plt.figure()
plt.plot(time,myarray1,label='t^2')
plt.plot(time,myarray2,label='2xt^2')
plt.legend()
plt.grid(True)
```
